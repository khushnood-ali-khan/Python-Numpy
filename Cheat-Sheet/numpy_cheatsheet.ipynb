# NumPy Cheat Sheet
A comprehensive, fully-executed reference for core NumPy operations — array creation, indexing, math, linear algebra, broadcasting, and more.

**Run all cells top to bottom** (or just read the outputs — they're already executed).

```python
import numpy as np
np.__version__
```

## 1. Array Creation

```python
# From lists
a = np.array([1, 2, 3])
b = np.array([[1, 2, 3], [4, 5, 6]])
print(a)
print(b)
print("shape:", b.shape, "| ndim:", b.ndim, "| dtype:", b.dtype, "| size:", b.size)
```

```python
# Built-in generators
print(np.zeros((2, 3)))
print(np.ones((2, 3)))
print(np.full((2, 2), 7))
print(np.eye(3))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
print(np.empty((2, 2)))  # uninitialized memory
```

```python
# Random arrays
rng = np.random.default_rng(42)  # preferred modern API (seeded for reproducibility)
print(rng.random((2, 3)))         # uniform [0,1)
print(rng.integers(0, 10, size=(2, 3)))  # random ints
print(rng.normal(0, 1, size=(2, 3)))     # normal distribution
```

## 2. Data Types (dtype)

```python
a = np.array([1, 2, 3], dtype=np.float32)
print(a, a.dtype)

b = a.astype(np.int64)
print(b, b.dtype)

print(np.array([1, 2, 3]).dtype)      # int64 (platform default)
print(np.array([1.0, 2, 3]).dtype)    # float64
print(np.array([1, 2, 3], dtype=complex))
```

## 3. Inspecting Arrays

```python
a = np.arange(24).reshape(4, 6)
print(a)
print("shape:", a.shape)
print("ndim:", a.ndim)
print("size:", a.size)
print("itemsize (bytes):", a.itemsize)
print("nbytes:", a.nbytes)
print("dtype:", a.dtype)
```

## 4. Indexing & Slicing

```python
a = np.arange(10)
print(a[2])          # single element
print(a[2:7])        # slice
print(a[::2])        # step
print(a[::-1])        # reverse
```

```python
m = np.arange(1, 13).reshape(3, 4)
print(m)
print(m[1, 2])        # row 1, col 2
print(m[0])            # first row
print(m[:, 1])         # second column
print(m[1:, 1:3])      # sub-matrix
print(m[-1])            # last row
```

```python
# Boolean (mask) indexing
a = np.array([1, -2, 3, -4, 5])
print(a[a > 0])
a[a < 0] = 0
print(a)
```

```python
# Fancy indexing
a = np.arange(10, 20)
idx = [0, 2, 4]
print(a[idx])

m = np.arange(12).reshape(3, 4)
print(m[[0, 2], [1, 3]])   # picks (0,1) and (2,3)
```

## 5. Reshaping & Manipulating Arrays

```python
a = np.arange(12)
print(a.reshape(3, 4))
print(a.reshape(3, -1))     # -1 infers dimension
print(a.reshape(3, 4).T)     # transpose
print(a.reshape(3, 4).flatten())  # copy
print(a.reshape(3, 4).ravel())    # view when possible
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(np.concatenate([a, b]))
print(np.stack([a, b]))          # new axis
print(np.vstack([a, b]))
print(np.hstack([a, b]))
print(np.split(np.arange(9), 3))
```

```python
a = np.arange(6)
print(np.expand_dims(a, axis=0).shape)  # (1, 6)
print(np.expand_dims(a, axis=1).shape)  # (6, 1)
print(a[:, np.newaxis].shape)            # same as above
print(np.squeeze(np.zeros((1, 3, 1))).shape)  # removes size-1 dims
```

## 6. Arithmetic & Universal Functions (ufuncs)

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print(a + b, a - b, a * b, a / b, a ** 2)
print(np.add(a, b), np.subtract(a, b), np.multiply(a, b))
```

```python
a = np.array([1, 4, 9, 16])
print(np.sqrt(a))
print(np.exp(np.array([0, 1, 2])))
print(np.log(np.array([1, np.e, np.e**2])))
print(np.abs(np.array([-3, -1, 2])))
print(np.round(np.array([1.234, 5.678]), 1))
print(np.sin(np.array([0, np.pi/2, np.pi])))
```

## 7. Aggregations & Statistics

```python
a = np.array([[1, 2, 3], [4, 5, 6]])
print("sum:", a.sum(), "| sum axis0:", a.sum(axis=0), "| sum axis1:", a.sum(axis=1))
print("mean:", a.mean())
print("std:", a.std(), "| var:", a.var())
print("min:", a.min(), "| max:", a.max())
print("argmin:", a.argmin(), "| argmax:", a.argmax())
print("cumsum:", a.cumsum())
print("median:", np.median(a))
print("percentile 90:", np.percentile(a, 90))
```

## 8. Broadcasting

```python
# Broadcasting lets NumPy operate on arrays of different shapes
a = np.array([[1, 2, 3], [4, 5, 6]])  # shape (2,3)
b = np.array([10, 20, 30])             # shape (3,)  -> broadcast to (2,3)
print(a + b)

c = np.array([[1], [2]])               # shape (2,1) -> broadcast to (2,3)
print(a + c)
```

## 9. Linear Algebra (`np.linalg`)

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("matmul:\n", A @ B)            # or np.matmul(A, B)
print("dot:\n", np.dot(A, B))
print("transpose:\n", A.T)
print("determinant:", np.linalg.det(A))
print("inverse:\n", np.linalg.inv(A))
print("trace:", np.trace(A))

eigvals, eigvecs = np.linalg.eig(A)
print("eigenvalues:", eigvals)
print("eigenvectors:\n", eigvecs)

# Solve Ax = b
b_vec = np.array([1, 2])
x = np.linalg.solve(A, b_vec)
print("solution x:", x)
```

## 10. Sorting & Searching

```python
a = np.array([3, 1, 4, 1, 5, 9, 2, 6])
print("sorted:", np.sort(a))
print("argsort:", np.argsort(a))
print("unique:", np.unique(a))
print("where >3:", np.where(a > 3))
print("values where >3:", a[np.where(a > 3)])
print("nonzero:", np.nonzero(a))
print("searchsorted:", np.searchsorted(np.sort(a), 4))
```

## 11. Set Operations

```python
a = np.array([1, 2, 3, 4])
b = np.array([3, 4, 5, 6])
print("union:", np.union1d(a, b))
print("intersect:", np.intersect1d(a, b))
print("diff:", np.setdiff1d(a, b))
print("in1d:", np.isin(a, b))
```

## 12. Copies vs Views

```python
a = np.arange(5)
view = a[1:4]        # view: shares memory
copy = a[1:4].copy()  # copy: independent memory

view[0] = 99
print("a after view edit:", a)   # a changes too

copy[0] = -1
print("a after copy edit:", a)   # a unaffected
print("np.shares_memory(a, view):", np.shares_memory(a, view))
print("np.shares_memory(a, copy):", np.shares_memory(a, copy))
```

## 13. Conditional Logic

```python
a = np.array([1, -2, 3, -4, 5])
print(np.where(a > 0, a, 0))          # replace negatives with 0
print(np.select([a < 0, a == 0, a > 0], ["neg", "zero", "pos"], default="unknown"))
print(np.clip(a, -1, 3))               # clamp values to range
```

## 14. Handling Missing / Special Values

```python
a = np.array([1, np.nan, 3, np.inf, -np.inf])
print(np.isnan(a))
print(np.isinf(a))
print(np.nan_to_num(a, nan=0.0, posinf=999, neginf=-999))
print(np.nansum(np.array([1, np.nan, 3])))
print(np.nanmean(np.array([1, np.nan, 3])))
```

## 15. Saving & Loading Arrays

```python
a = np.arange(10)
np.save('/tmp/array.npy', a)
loaded = np.load('/tmp/array.npy')
print(loaded)

np.savetxt('/tmp/array.csv', a, delimiter=',')
loaded_csv = np.loadtxt('/tmp/array.csv', delimiter=',')
print(loaded_csv)
```

## 16. Useful Random / Sampling Recipes

```python
rng = np.random.default_rng(0)

print(rng.choice([10, 20, 30, 40], size=3, replace=False))  # sample without replacement
print(rng.permutation(np.arange(5)))                          # shuffle a copy
a = np.arange(5)
rng.shuffle(a)                                                  # shuffle in place
print(a)
```
