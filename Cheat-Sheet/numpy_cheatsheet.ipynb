# NumPy Cheat Sheet
A comprehensive, fully-executed reference for core NumPy operations — array creation, indexing, math, linear algebra, broadcasting, and more. Every line has a comment explaining its use.

**Run all cells top to bottom** (or just read the outputs — they're already executed).

```python
import numpy as np  # import numpy under the standard alias
np.__version__  # check installed numpy version
```

## 1. Array Creation

```python
a = np.array([1, 2, 3])  # create a 1D array from a list
b = np.array([[1, 2, 3], [4, 5, 6]])  # create a 2D array from a nested list
print(a)  # display 1D array
print(b)  # display 2D array
print("shape:", b.shape, "| ndim:", b.ndim, "| dtype:", b.dtype, "| size:", b.size)  # inspect array metadata
```

```python
print(np.zeros((2, 3)))  # array filled with zeros
print(np.ones((2, 3)))  # array filled with ones
print(np.full((2, 2), 7))  # array filled with a constant value
print(np.eye(3))  # identity matrix
print(np.arange(0, 10, 2))  # evenly spaced values by step size
print(np.linspace(0, 1, 5))  # evenly spaced values by count
print(np.empty((2, 2)))  # uninitialized array (fast, garbage values)
```

```python
rng = np.random.default_rng(42)  # create a seeded random generator (reproducible)
print(rng.random((2, 3)))  # uniform random floats in [0,1)
print(rng.integers(0, 10, size=(2, 3)))  # random integers in a range
print(rng.normal(0, 1, size=(2, 3)))  # random samples from normal distribution
```

## 2. Data Types (dtype)

```python
a = np.array([1, 2, 3], dtype=np.float32)  # create array with explicit dtype
print(a, a.dtype)  # show array and its dtype

b = a.astype(np.int64)  # cast array to a different dtype
print(b, b.dtype)  # show casted array and new dtype

print(np.array([1, 2, 3]).dtype)  # default int dtype for integer literals
print(np.array([1.0, 2, 3]).dtype)  # default float dtype when a float is present
print(np.array([1, 2, 3], dtype=complex))  # force complex dtype
```

## 3. Inspecting Arrays

```python
a = np.arange(24).reshape(4, 6)  # create a 4x6 array of sequential values
print(a)  # display the array
print("shape:", a.shape)  # dimensions of the array
print("ndim:", a.ndim)  # number of dimensions
print("size:", a.size)  # total number of elements
print("itemsize (bytes):", a.itemsize)  # bytes per element
print("nbytes:", a.nbytes)  # total bytes consumed by the array
print("dtype:", a.dtype)  # data type of elements
```

## 4. Indexing & Slicing

```python
a = np.arange(10)  # create a 1D array 0-9
print(a[2])  # single element access
print(a[2:7])  # slice a range of elements
print(a[::2])  # slice with a step (every 2nd element)
print(a[::-1])  # reverse the array
```

```python
m = np.arange(1, 13).reshape(3, 4)  # create a 3x4 matrix
print(m)  # display the matrix
print(m[1, 2])  # access element at row 1, col 2
print(m[0])  # access first row
print(m[:, 1])  # access second column (all rows)
print(m[1:, 1:3])  # access a sub-matrix slice
print(m[-1])  # access last row
```

```python
a = np.array([1, -2, 3, -4, 5])  # array with positive and negative values
print(a[a > 0])  # boolean mask: select only positive elements
a[a < 0] = 0  # boolean mask assignment: zero out negative elements
print(a)  # display modified array
```

```python
a = np.arange(10, 20)  # array of values 10-19
idx = [0, 2, 4]  # list of indices to select
print(a[idx])  # fancy indexing: select elements at given indices

m = np.arange(12).reshape(3, 4)  # create a 3x4 matrix
print(m[[0, 2], [1, 3]])  # fancy indexing: picks (0,1) and (2,3)
```

## 5. Reshaping & Manipulating Arrays

```python
a = np.arange(12)  # 1D array of 12 elements
print(a.reshape(3, 4))  # reshape into 3x4 matrix
print(a.reshape(3, -1))  # reshape with -1 to auto-infer that dimension
print(a.reshape(3, 4).T)  # transpose rows and columns
print(a.reshape(3, 4).flatten())  # flatten to 1D, returns a copy
print(a.reshape(3, 4).ravel())  # flatten to 1D, returns a view when possible
```

```python
a = np.array([1, 2, 3])  # first array
b = np.array([4, 5, 6])  # second array
print(np.concatenate([a, b]))  # join arrays along an existing axis
print(np.stack([a, b]))  # join arrays along a new axis
print(np.vstack([a, b]))  # stack arrays vertically (row-wise)
print(np.hstack([a, b]))  # stack arrays horizontally (column-wise)
print(np.split(np.arange(9), 3))  # split an array into equal parts
```

```python
a = np.arange(6)  # 1D array of 6 elements
print(np.expand_dims(a, axis=0).shape)  # add a new axis at position 0 -> (1, 6)
print(np.expand_dims(a, axis=1).shape)  # add a new axis at position 1 -> (6, 1)
print(a[:, np.newaxis].shape)  # same effect using np.newaxis
print(np.squeeze(np.zeros((1, 3, 1))).shape)  # remove all size-1 dimensions
```

## 6. Arithmetic & Universal Functions (ufuncs)

```python
a = np.array([1, 2, 3])  # first array
b = np.array([4, 5, 6])  # second array
print(a + b, a - b, a * b, a / b, a ** 2)  # elementwise arithmetic operators
print(np.add(a, b), np.subtract(a, b), np.multiply(a, b))  # equivalent ufunc calls
```

```python
a = np.array([1, 4, 9, 16])  # array of perfect squares
print(np.sqrt(a))  # elementwise square root
print(np.exp(np.array([0, 1, 2])))  # elementwise exponential (e^x)
print(np.log(np.array([1, np.e, np.e**2])))  # elementwise natural log
print(np.abs(np.array([-3, -1, 2])))  # elementwise absolute value
print(np.round(np.array([1.234, 5.678]), 1))  # round to given decimal places
print(np.sin(np.array([0, np.pi/2, np.pi])))  # elementwise sine
```

## 7. Aggregations & Statistics

```python
a = np.array([[1, 2, 3], [4, 5, 6]])  # 2D array for aggregation examples
print("sum:", a.sum(), "| sum axis0:", a.sum(axis=0), "| sum axis1:", a.sum(axis=1))  # totals overall and per axis
print("mean:", a.mean())  # average of all elements
print("std:", a.std(), "| var:", a.var())  # standard deviation and variance
print("min:", a.min(), "| max:", a.max())  # smallest and largest values
print("argmin:", a.argmin(), "| argmax:", a.argmax())  # index of smallest/largest value
print("cumsum:", a.cumsum())  # running cumulative sum
print("median:", np.median(a))  # median value
print("percentile 90:", np.percentile(a, 90))  # 90th percentile value
```

## 8. Broadcasting

```python
a = np.array([[1, 2, 3], [4, 5, 6]])  # shape (2,3) array
b = np.array([10, 20, 30])  # shape (3,) array, broadcasts across rows of a
print(a + b)  # broadcasting: b is applied to every row of a

c = np.array([[1], [2]])  # shape (2,1) array, broadcasts across columns of a
print(a + c)  # broadcasting: c is applied to every column of a
```

## 9. Linear Algebra (`np.linalg`)

```python
A = np.array([[1, 2], [3, 4]])  # first matrix
B = np.array([[5, 6], [7, 8]])  # second matrix

print("matmul:\n", A @ B)  # matrix multiplication (or np.matmul(A, B))
print("dot:\n", np.dot(A, B))  # dot product (same as matmul for 2D arrays)
print("transpose:\n", A.T)  # transpose of matrix A
print("determinant:", np.linalg.det(A))  # determinant of matrix A
print("inverse:\n", np.linalg.inv(A))  # inverse of matrix A
print("trace:", np.trace(A))  # sum of diagonal elements

eigvals, eigvecs = np.linalg.eig(A)  # compute eigenvalues and eigenvectors
print("eigenvalues:", eigvals)  # display eigenvalues
print("eigenvectors:\n", eigvecs)  # display eigenvectors

b_vec = np.array([1, 2])  # right-hand side vector for Ax = b
x = np.linalg.solve(A, b_vec)  # solve the linear system Ax = b
print("solution x:", x)  # display solution vector
```

## 10. Sorting & Searching

```python
a = np.array([3, 1, 4, 1, 5, 9, 2, 6])  # unsorted array with duplicates
print("sorted:", np.sort(a))  # return a sorted copy
print("argsort:", np.argsort(a))  # indices that would sort the array
print("unique:", np.unique(a))  # unique values, sorted
print("where >3:", np.where(a > 3))  # indices where condition is true
print("values where >3:", a[np.where(a > 3)])  # values matching condition
print("nonzero:", np.nonzero(a))  # indices of nonzero elements
print("searchsorted:", np.searchsorted(np.sort(a), 4))  # insertion index to keep array sorted
```

## 11. Set Operations

```python
a = np.array([1, 2, 3, 4])  # first set of values
b = np.array([3, 4, 5, 6])  # second set of values
print("union:", np.union1d(a, b))  # all unique values from both arrays
print("intersect:", np.intersect1d(a, b))  # values common to both arrays
print("diff:", np.setdiff1d(a, b))  # values in a but not in b
print("in1d:", np.isin(a, b))  # boolean mask of a's elements found in b
```

## 12. Copies vs Views

```python
a = np.arange(5)  # base array
view = a[1:4]  # slicing creates a view (shares memory with a)
copy = a[1:4].copy()  # .copy() creates an independent array

view[0] = 99  # modify the view
print("a after view edit:", a)  # a changes too since view shares memory

copy[0] = -1  # modify the copy
print("a after copy edit:", a)  # a unaffected since copy is independent
print("np.shares_memory(a, view):", np.shares_memory(a, view))  # confirm view shares memory with a
print("np.shares_memory(a, copy):", np.shares_memory(a, copy))  # confirm copy does not share memory
```

## 13. Conditional Logic

```python
a = np.array([1, -2, 3, -4, 5])  # array with mixed sign values
print(np.where(a > 0, a, 0))  # replace negative values with 0, keep positives
print(np.select([a < 0, a == 0, a > 0], ["neg", "zero", "pos"], default="unknown"))  # multi-condition labeling
print(np.clip(a, -1, 3))  # clamp all values into the range [-1, 3]
```

## 14. Handling Missing / Special Values

```python
a = np.array([1, np.nan, 3, np.inf, -np.inf])  # array with NaN and infinities
print(np.isnan(a))  # boolean mask of NaN positions
print(np.isinf(a))  # boolean mask of infinite positions
print(np.nan_to_num(a, nan=0.0, posinf=999, neginf=-999))  # replace NaN/inf with finite values
print(np.nansum(np.array([1, np.nan, 3])))  # sum ignoring NaN values
print(np.nanmean(np.array([1, np.nan, 3])))  # mean ignoring NaN values
```

## 15. Saving & Loading Arrays

```python
a = np.arange(10)  # array to persist
np.save('/tmp/array.npy', a)  # save array to binary .npy file
loaded = np.load('/tmp/array.npy')  # load array back from .npy file
print(loaded)  # display loaded array

np.savetxt('/tmp/array.csv', a, delimiter=',')  # save array to a text/CSV file
loaded_csv = np.loadtxt('/tmp/array.csv', delimiter=',')  # load array back from CSV file
print(loaded_csv)  # display loaded array
```

## 16. Useful Random / Sampling Recipes

```python
rng = np.random.default_rng(0)  # seeded random generator for reproducibility

print(rng.choice([10, 20, 30, 40], size=3, replace=False))  # random sample without replacement
print(rng.permutation(np.arange(5)))  # return a shuffled copy of the array
a = np.arange(5)  # array to shuffle in place
rng.shuffle(a)  # shuffle the array in place (modifies original)
print(a)  # display shuffled array
```
